```python
import qiskit
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram

print("Qiskit Version:", qiskit.__version__)
simulator = AerSimulator()
print(simulator)
```

```python
# Simulated unknown rotation angle
true_theta = 1.20
shots = 10000

qc = QuantumCircuit(1, 1)
qc.ry(true_theta, 0)
qc.measure(0, 0)

counts = simulator.run(qc, shots=shots).result().get_counts()

p1 = counts.get("1", 0) / shots
estimated_theta = 2 * np.arcsin(np.sqrt(p1))

print("Measurement counts:", counts)
print(f"Estimated P(1): {p1:.4f}")
print(f"True angle: {true_theta:.4f} radians")
print(f"Estimated angle: {estimated_theta:.4f} radians")
print(f"Absolute error: {abs(true_theta-estimated_theta):.4f} radians")
```
