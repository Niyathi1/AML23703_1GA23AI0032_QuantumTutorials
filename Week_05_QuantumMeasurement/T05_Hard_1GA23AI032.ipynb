```python
import qiskit
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram

print("Qiskit Version:", qiskit.__version__)
simulator = AerSimulator()
print(simulator)
```

```python
# Prepare Bell pair
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)

bell_state = Statevector.from_instruction(bell)
print("Bell pair statevector:")
print(bell_state)

# Measure only qubit 0
partial_measure = QuantumCircuit(2, 1)
partial_measure.h(0)
partial_measure.cx(0, 1)
partial_measure.measure(0, 0)

print("\nPartial measurement circuit:")
print(partial_measure)

counts = simulator.run(partial_measure, shots=1024).result().get_counts()
print("\nMeasurement results for qubit 0:", counts)

plot_histogram(counts, title="Partial Measurement of Bell Pair")

print("\nIf measured qubit is 0, the other qubit collapses to |0>.")
print("If measured qubit is 1, the other qubit collapses to |1>.")
```
