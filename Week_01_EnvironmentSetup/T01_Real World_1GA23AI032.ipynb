```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import random

sim = AerSimulator()

quantum = []

# Generate 10 quantum random bits
for _ in range(10):
    qc = QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)

    job = sim.run(transpile(qc, sim), shots=1)
    bit = list(job.result().get_counts().keys())[0]
    quantum.append(int(bit))

# Python random bits
python = [random.randint(0,1) for _ in range(10)]

print("Quantum Random:", quantum)
print("Python Random :", python)
```
