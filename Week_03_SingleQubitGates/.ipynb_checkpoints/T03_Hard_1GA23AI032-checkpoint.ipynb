```python
# Import Required Libraries
import qiskit
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram, plot_bloch_multivector

print("Qiskit Version:", qiskit.__version__)
simulator = AerSimulator()
print(simulator)
```

```python
# Five randomly chosen single-qubit gates

import random

gate_names = ["X", "Y", "Z", "H", "S", "T"]
random.seed(42)  # fixed seed so the sequence is reproducible

selected_gates = random.choices(gate_names, k=5)
print("Selected gate sequence:", selected_gates)

qc5 = QuantumCircuit(1)

# Matrix representation of the gates
I = np.eye(2, dtype=complex)
X = np.array([[0, 1], [1, 0]], dtype=complex)
Y = np.array([[0, -1j], [1j, 0]], dtype=complex)
Z = np.array([[1, 0], [0, -1]], dtype=complex)
H = (1 / np.sqrt(2)) * np.array([[1, 1], [1, -1]], dtype=complex)
S = np.array([[1, 0], [0, 1j]], dtype=complex)
T = np.array([[1, 0], [0, np.exp(1j*np.pi/4)]], dtype=complex)

matrix_map = {"X": X, "Y": Y, "Z": Z, "H": H, "S": S, "T": T}

# Initial state |0>
analytic_state = np.array([1, 0], dtype=complex)

for g in selected_gates:
    getattr(qc5, g.lower())(0)
    analytic_state = matrix_map[g] @ analytic_state

print("\nCircuit:")
print(qc5)

print("\nAnalytical final state:")
print(analytic_state)

sim_state = Statevector.from_instruction(qc5).data
print("\nSimulation final state:")
print(sim_state)

print("\nAnalytical and simulation results match:",
      np.allclose(analytic_state, sim_state))
```
