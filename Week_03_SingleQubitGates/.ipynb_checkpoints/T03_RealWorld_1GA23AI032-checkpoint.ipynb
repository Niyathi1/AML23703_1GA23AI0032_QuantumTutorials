```python
# Import Required Libraries
import qiskit
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram, plot_bloch_multivector

print("Qiskit Version:", qiskit.__version__)
simulator = AerSimulator()
print(simulator)
```

```python
# Quantum coin-flip game
# H creates an equal superposition. Measurement decides the outcome.

qc_coin = QuantumCircuit(1, 1)
qc_coin.h(0)
qc_coin.measure(0, 0)

print(qc_coin)

job = simulator.run(qc_coin, shots=4096)
result = job.result()
counts = result.get_counts()

print("Coin-flip results:", counts)

p0 = counts.get("0", 0) / 4096
p1 = counts.get("1", 0) / 4096

print(f"Probability of outcome 0 (Heads): {p0:.3f}")
print(f"Probability of outcome 1 (Tails): {p1:.3f}")

plot_histogram(counts, title="Quantum Coin-Flip Probability")
```
