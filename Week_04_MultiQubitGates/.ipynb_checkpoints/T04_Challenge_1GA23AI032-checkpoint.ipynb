```python
# Import Required Libraries
import qiskit
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector, partial_trace, entropy
from qiskit.visualization import plot_histogram

print("Qiskit Version:", qiskit.__version__)
simulator = AerSimulator()
print(simulator)
```

```python
# Arbitrary 2-qubit entangled state
# |psi> = cos(theta/2)|00> + sin(theta/2)|11>
# with theta = pi/3, giving unequal amplitudes.

theta = np.pi / 3

arbitrary = QuantumCircuit(2)
arbitrary.ry(theta, 0)
arbitrary.cx(0, 1)

print("Arbitrary entangled-state circuit:")
print(arbitrary)

state = Statevector.from_instruction(arbitrary)

print("\nStatevector:")
print(state)

print("\nState probabilities:")
for i, probability in enumerate(np.abs(state.data) ** 2):
    print(f"|{i:02b}> : {probability:.4f}")

# Reduced-state entropy verifies entanglement
reduced_q0 = partial_trace(state, [1])
entanglement_entropy = entropy(reduced_q0)

print(f"\nEntanglement entropy of qubit 0: {entanglement_entropy:.4f}")
print("Entangled state:", entanglement_entropy > 0)
```
