```python
# Import Required Libraries
import qiskit
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector, partial_trace, entropy
from qiskit.visualization import plot_histogram

print("Qiskit Version:", qiskit.__version__)
simulator = AerSimulator()
print(simulator)
```

```python
# Classical full-adder using Toffoli and CNOT gates
# q0 = A, q1 = B, q2 = Cin, q3 = Carry
# q2 becomes Sum

def full_adder_circuit(A, B, Cin):
    qc = QuantumCircuit(4, 4)

    if A:
        qc.x(0)
    if B:
        qc.x(1)
    if Cin:
        qc.x(2)

    # Sum = A XOR B XOR Cin
    qc.cx(0, 2)
    qc.cx(1, 2)

    # Carry = AB OR Cin(A XOR B)
    qc.ccx(0, 1, 3)
    qc.cx(0, 1)
    qc.ccx(1, 2, 3)
    qc.cx(0, 1)

    qc.measure([0, 1, 2, 3], [0, 1, 2, 3])
    return qc

print("A B Cin | Sum Carry")
print("-------------------")

for A in [0, 1]:
    for B in [0, 1]:
        for Cin in [0, 1]:
            qc = full_adder_circuit(A, B, Cin)
            counts = simulator.run(qc, shots=256).result().get_counts()
            measured = next(iter(counts))

            # Display mapping: c3 c2 c1 c0
            sum_bit = int(measured[1])
            carry_bit = int(measured[0])

            print(f"{A} {B}  {Cin}  |  {sum_bit}    {carry_bit}")
```
